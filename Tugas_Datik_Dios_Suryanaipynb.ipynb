```python
import kagglehub
import pandas as pd
import os

path = kagglehub.dataset_download("codebynadiia/gdp-per-country-20152025")

print("Dataset tersimpan di:", path)

print(os.listdir(path))
```

```python
# Install dependensi

import kagglehub
from kagglehub import KaggleDatasetAdapter

file_path = "2015-2025.csv"

# muat versi akhirnya
df = kagglehub.load_dataset(
  KaggleDatasetAdapter.PANDAS,
  "codebynadiia/gdp-per-country-20152025",
  file_path,

)

print("First 5 records:", df.head())
```

```python
file_path = os.path.join(path, "2015-2025.csv")

df = pd.read_csv(file_path)

print(df.head())
print(df.columns)
```

```python
#Cleaning datanya

print("Jumlah baris dan kolom:", df.shape)
print("\nNama kolom:")
print(df.columns.tolist())

# Hapus spasi pada nama negara
df["Country"] = df["Country"].str.strip()
years = [str(year) for year in range(2015, 2025)]

for year in years:
    df[year] = pd.to_numeric(df[year], errors="coerce")

# Cek missing value
print("\nMissing value:")
print(df[["Country"] + years].isnull().sum())

# Hapus nama negara yg kosoong dan duplikat

df = df.dropna(subset=["Country"])
df = df.drop_duplicates(subset=["Country"])

print("\nUkuran data setelah cleaning:", df.shape)
```

```python
#saya pilih 5 negara
countries = [
    "United States",
    "China",
    "India",
    "Japan",
    "Germany"
]

df_selected = df[df["Country"].isin(countries)].copy()

df_selected[["Country", "2015", "2016", "2017", "2018", "2019", "2020", "2021", "2022", "2023", "2024"]]
```

```python
#ngubah formatnya agar kebawah
years = [str(year) for year in range(2015, 2025)]

df_long = df_selected.melt(
    id_vars="Country",
    value_vars=years,
    var_name="Year",
    value_name="GDP"
)

df_long["Year"] = df_long["Year"].astype(int)

df_long.head(10)
```

```python
df_race = df_long.pivot(
    index="Year",
    columns="Country",
    values="GDP"
)

df_race
```
